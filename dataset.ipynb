-Roboflow : 1200 imagenes (Train+valid +test)

-Soccernet: 100 clips compuesto por 750 imganes cada una

```python
import os
import dataset
import importlib
import cv2
import matplotlib.pyplot as plt
import random
import shutil
importlib.reload(dataset)
```

```python
train_path = "Dataset/detection/train"
test_path = "Dataset/detection/test"
valid_path = "Dataset/detection/valid"

os.makedirs(train_path)
os.makedirs(os.path.join(train_path,"images"))
os.makedirs(os.path.join(train_path,"labels"))
os.makedirs(test_path)
os.makedirs(os.path.join(test_path,"images"))
os.makedirs(os.path.join(test_path,"labels"))
os.makedirs(valid_path)
os.makedirs(os.path.join(valid_path,"images"))
os.makedirs(os.path.join(valid_path,"labels"))
```

```python
#Copiar y renombrar imagenes de roboflow
num_img = 1
path = "roboflow/valid"
imgs_path = os.path.join(path,"images")
labels_path = os.path.join(path,"labels")

imgs = os.listdir(imgs_path)
for img in imgs:
    #Imagen
    img_path = os.path.join(imgs_path,img)
    img_new_path =f"Dataset/detection/valid/images/{num_img:06d}.jpg"
    shutil.copy(img_path, img_new_path)
    #Etiqueta
    label = os.path.splitext(img)[0]+".txt"
    label_path = os.path.join(labels_path,label)
    label_new_path = f"Dataset/detection/valid/labels/{num_img:06d}.txt"
    shutil.copy(label_path, label_new_path)

    num_img = num_img+1
```

```python
#Seleccionar y copiar imagenes de Soccernet

path = "SoccerNet/tracking/test_sn"
num_img = len(os.listdir("Dataset/detection/test/images")) +1 

#SoccerNet imgs
folders = os.listdir(path)

for folder in folders:
    imgs_path = os.path.join(path,folder,"img1")
    labels_path = os.path.join(path,folder,"labels")

    imgs = os.listdir(imgs_path)
    random_sample = random.sample(imgs,10)

    for img in random_sample:
        #Imagen
        img_path = os.path.join(imgs_path,img)
        img_new_path =f"Dataset/detection/test/images/{num_img:06d}.jpg"
        shutil.copy(img_path, img_new_path)
        #Etiqueta
        label = img.split(".")[0]+".txt"
        label_path = os.path.join(labels_path,label)
        label_new_path = f"Dataset/detection/test/labels/{num_img:06d}.txt"
        shutil.copy(label_path, label_new_path)

        num_img = num_img+1
```
