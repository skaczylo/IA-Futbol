# Dataset Engineering

En este notebook se analizarán distintas formas de generar nuevos datos etiquetados con el objetivo de ampliar un dataset pequeño o crear uno desde cero y poder entrenar un modelo con dichos datos.

Nuestro dataset estará formada por imagenes de partidos de fútbol y necesitaremos que en cada imagen estén marcados los siguientes objetos:

- Balón
- Jugadores
- Árbitros

Uno de los formatos más usados para el marcaje de objetos es el denominado **bounding box** : un rectángulo que define el área mínima que rodea completamente el objeto de interés

### Obtención Imagenes


```python
path = "Dataset/detection/no label/images/53.jpg"
```

```python
#codigo para marcar imagenes
import cv2
import numpy as np
img_path = "dataset_engineering/235.png"
img = cv2.imread(img_path)



# Definir rango para verdes
lower_green = np.array([40, 80, 50])
upper_green = np.array([80, 255, 255])

# Convertir a HSV
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Crear máscara para verde
mask = cv2.inRange(img_hsv, lower_green, upper_green)

# Invertir máscara
mask_in = cv2.bitwise_not(mask)

# Encontrar contornos
contours, hierarchy = cv2.findContours(mask_in, cv2.RETR_CCOMP, cv2.CHAIN_APPROX_SIMPLE)

# Dibujar contornos en verde
for cnt in contours:

    if cv2.contourArea(cnt) > 100:
        x1,y1,w,h = cv2.boundingRect(cnt)
        cv2.rectangle(img, (x1, y1), (x1 + w, y1 + h), (0, 255, 0), 2)
        print(cv2.contourArea(cnt))
    

# Reducir tamaño de imagen para mostrar (opcional)
img = cv2.resize(img, (0, 0), fx=0.5, fy=0.5)

# Mostrar máscara invertida (para debug)
cv2.imshow("mask_in", mask_in)
cv2.imshow("img", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
